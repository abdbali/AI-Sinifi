```python
import gradio as gr

print("Gradio kütüphanesi başarıyla yüklendi.")
```

## Adım 4: En Basit Çekirdek Fonksiyonu (Backend) Yazma

Gradio'nun mantığı şudur: O sizden her zaman bir fonksiyon (`def`) bekler. Girdileri alır, o fonksiyona sokar, fonksiyonun `return` (geri döndürdüğü) değerini ekrana yansıtır.

```python
# Kullanıcıdan ismini alıp ona hoş geldin diyen basit bir fonksiyon
def selamla(kullanici_ismi):
    cevap = f"Merhaba {kullanici_ismi}! Kendi yazdığın ilk web sitene hoş geldin."
    return cevap

# Bu fonksiyon normalde sadece print(selamla("Ali")) ile çalışırdı.
```

## Adım 5: Fonksiyonu Web Sitesine (Arayüze) Çevirme

`gr.Interface` (Gradio Arayüz) komutu 3 temel parametre alır:
1. `fn`: Çalıştırılacak fonksiyonun adı.
2. `inputs`: Kullanıcının veri gireceği kutunun türü (Örn: "text" yani metin kutusu)
3. `outputs`: Sonucun gösterileceği kutunun türü (Örn: "text")

```python
# Arayüzü tanımlıyoruz (Henüz çalıştırmadık)
arayuz = gr.Interface(
    fn=selamla,
    inputs="text",
    outputs="text"
)
print("Arayüz tasarımı tamamlandı.")
```

## Adım 6: Arayüzü Çalıştırma (Launch) ve Public Link Alma

Web sitemizi başlatalım. Çıkan hücrenin altında (Colab içinde) siteniz belirecektir. Ayrıca kod size `gradio.live` ile biten bir link verecek, o linke tıklayarak sitenizi yeni sekmede açabilirsiniz!

```python
# share=True komutu, bu sitenin size özel geçici (72 saatlik) bir internet linkini oluşturur.
arayuz.launch(share=True)
```

## Adım 7: Girdi Tiplerini Öğrenme (Slider - Kaydırma Çubuğu)

Sadece metin değil, bir kaydırma çubuğu (Slider) ile yaş hesaplayan yeni bir site yapalım.

```python
def yas_hesapla(dogum_yili):
    yas = 2026 - dogum_yili
    return f"Siz tam {yas} yaşındasınız!"

# inputs kısmına gr.Slider(minimum, maksimum) veriyoruz
yas_arayuzu = gr.Interface(
    fn=yas_hesapla,
    inputs=gr.Slider(1900, 2026, step=1, label="Doğum Yılınızı Seçin"),
    outputs="text"
)

yas_arayuzu.launch(share=True)
```

## Adım 8: Daha Şık Bir Tasarım (Başlık ve Açıklama)

Profesyonel görünmesi için siteye Başlık (title) ve Açıklama (description) ekleyebiliriz.

```python
sik_arayuz = gr.Interface(
    fn=yas_hesapla,
    inputs=gr.Slider(1900, 2026, step=1, label="Doğum Yılı"),
    outputs=gr.Textbox(label="Yapay Zeka Hesabı"), # Çıktı kutusuna da isim verdik
    title="Zaman Yolcusu Yaş Hesaplayıcı",
    description="Lütfen aşağıdaki çubuğu kaydırarak doğum yılınızı seçin, sistem yaşınızı bulsun."
)

sik_arayuz.launch(share=True)
```
