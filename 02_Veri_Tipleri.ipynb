## Adım 2: Teknik Terimler ve Sözlük

- Integer (int): Tam sayılar. Küsuratı olmayan eksi veya artı yöndeki sayılar (Örn: -5, 0, 42).
- Float: Ondalıklı (virgüllü) sayılar. Bilgisayar dilinde virgül yerine nokta kullanılır (Örn: 3.14, 2.5).
- String (str): Metin veya karakter dizileri. Her zaman tırnak işaretleri arasında yazılırlar.
- Boolean (bool): Mantıksal doğruluk değerleri. Yalnızca iki durum alabilir: True (Doğru) veya False (Yanlış).
- Type Casting: Bir veri tipini zorla başka bir veri tipine çevirme işlemidir (Örneğin bir metni sayıya çevirmek).
- İndeks (Index): Bir metnin içindeki harflerin veya bir listenin içindeki elemanların sıra numarasıdır. Programlamada sayma işlemi her zaman 0'dan başlar.

## Adım 3: Tam Sayılar (Integer) ve Ondalıklı Sayılar (Float)

Matematiksel işlemlerde kullandığımız sayılar iki ana gruba ayrılır. Aşağıdaki hücrede tam sayılar ve ondalıklı sayılar arasındaki farkları ve bu farkların bellekte nasıl saklandığını göreceğiz.

```python
# Tam sayı (Integer) tanımlamaları
ogrenci_sayisi = 25
hava_sicakligi = -5

# Ondalıklı sayı (Float) tanımlamaları
# Dikkat: Programlamada küsuratlar virgül (,) ile değil nokta (.) ile ayrılır!
pi_degeri = 3.14159
boy_uzunlugu = 1.75

print("Öğrenci Sayısı:", ogrenci_sayisi)
print("Pi Değeri:", pi_degeri)

# İki farklı sayı tipini matematiksel olarak işleyebiliriz
toplam = ogrenci_sayisi + boy_uzunlugu
print("Bir int ile bir float toplandığında sonuç her zaman kapsayıcı olan float türünde olur:", toplam)
```

## Adım 4: Metin Veri Tipi (String) ve İndeksleme

String veri tipi tek tırnak ('...'), çift tırnak ("...") veya çok satırlı metinler için üç tırnak ("""...""") arasında yazılır. Bir string aslında karakterlerin yan yana gelmiş halidir (Dizi - Array).

```python
okul_ismi = "Atatürk Lisesi"
ders_adi = 'Yapay Zeka Uygulamaları'

# Metinleri birleştirmek (Concatenation)
tam_metin = okul_ismi + " - " + ders_adi
print("Birleştirilmiş Metin:", tam_metin)

# İndeksleme: Bilgisayarlar saymaya 0'dan başlar!
# 'okul_ismi' değişkenindeki ilk harfi almak için köşeli parantez kullanırız:
ilk_harf = okul_ismi[0]
print("Okul isminin ilk harfi:", ilk_harf)

# 4. harfi almak için indeks olarak 3 kullanmalıyız (0, 1, 2, 3)
dorduncu_harf = okul_ismi[3]
print("Okul isminin dördüncü harfi:", dorduncu_harf)
```

## Adım 5: Metin Düzenleme Metotları (String Methods)

Yapay zeka modelleri genellikle metinleri anlamadan önce temizler ve düzenler. Python, stringler üzerinde işlem yapmamız için bize hazır araçlar (metotlar) sunar.

```python
ornek_cumle = "Bu Cümle İçerisinde FARKLI büyüklükte HArfler VAR."

# .lower() metodu: Tüm harfleri küçültür.
kucuk_cumle = ornek_cumle.lower()
print("Tamamen Küçük:", kucuk_cumle)

# .upper() metodu: Tüm harfleri büyütür.
buyuk_cumle = ornek_cumle.upper()
print("Tamamen Büyük:", buyuk_cumle)

# .replace(eski, yeni) metodu: Bir kelimeyi veya harfi başka bir kelimeyle değiştirir.
degistirilmis_cumle = ornek_cumle.replace("FARKLI", "BENZER")
print("Değiştirilmiş Metin:", degistirilmis_cumle)

# len() fonksiyonu (Length): Metnin (boşluklar dahil) kaç karakterden oluştuğunu sayar.
karakter_sayisi = len(ornek_cumle)
print("Bu cümlenin toplam uzunluğu:", karakter_sayisi)
```

## Adım 6: Mantıksal Veri Tipi (Boolean)

Boolean veri tipi programlamanın karar verme mekanizmasıdır. Sadece 'True' veya 'False' değerini alabilir. Genellikle eşitlik kontrolü sonucunda ortaya çıkar.

```python
sistem_acik_mi = True
hata_var_mi = False

print("Sistem açık mı?", sistem_acik_mi)

# Karşılaştırma operatörleri otomatik olarak Boolean sonuç üretir:
# '==' operatörü 'iki değer eşit midir?' anlamına gelir.
# '>=' operatörü 'büyük veya eşit midir?' anlamına gelir.

yas = 18
resit_mi = yas >= 18
print("Kişi reşit mi? (Yaş 18'den büyük veya eşit mi?)", resit_mi)

isim_kontrol = ("Ahmet" == "Mehmet")
print("Ahmet ile Mehmet aynı kelimeler mi?", isim_kontrol)
```

## Adım 7: Değişken Tiplerini Analiz Etme (type fonksiyonu)

Bir değişkenin içerdiği verinin tipinden emin değilsek, `type()` fonksiyonunu kullanarak bilgisayara bunu sorabiliriz.

```python
veri_1 = 1923
veri_2 = "1923"
veri_3 = 3.14
veri_4 = True

# type() fonksiyonu verinin sınıfını (class) döndürür.
print("veri_1'in tipi:", type(veri_1))
print("veri_2'nin tipi:", type(veri_2))
print("veri_3'ün tipi:", type(veri_3))
print("veri_4'ün tipi:", type(veri_4))
```

## Adım 8: Veri Tipi Dönüşümleri (Type Casting)

Bir sayıyı metne dönüştürüp yazdırmak isteyebiliriz veya metin formunda (string) aldığımız bir sayıyı matematiksel işleme sokmak için gerçek sayı tipine (integer) çevirmemiz gerekebilir.

```python
# Metin (String) olarak tanımlanmış sayılar
sayi_metni = "50"

# Eğer sayi_metni ile matematiksel işlem yaparsak beklediğimiz sayısal sonucu alamayız.
# '50' metnini 2 ile çarpmak, 50 metnini iki kere yan yana yazdırmak demektir.
print("Metin olarak çarpma sonucu:", sayi_metni * 2) # Sonuç: 5050

# Metni tam sayıya (Integer) çevirmek için int() fonksiyonu kullanılır.
gercek_sayi = int(sayi_metni)
print("Dönüştürüldükten sonra çarpma sonucu:", gercek_sayi * 2) # Sonuç: 100

# Tam tersi işlem: Sayıyı metne çevirmek için str() fonksiyonu kullanılır.
oyuncu_puani = 2500
rapor_cumlesi = "Oyuncunun bu maçtaki toplam puanı: " + str(oyuncu_puani)
print(rapor_cumlesi)
```

## Adım 9: Dinamik Metin Formatlama (F-String)

Değişkenleri metinlerin içine yerleştirmek için sürekli `+ str()` kullanmak kodun okunabilirliğini düşürür. Python'daki 'f-string' yapısı sayesinde metinlerin içine değişkenleri doğrudan, daha temiz bir biçimde gömebiliriz.

```python
isim = "Zeynep"
okul = "Bilim Lisesi"
not_ortalamasi = 95.5

# Metnin en başına 'f' harfi koyarak içeride süslü parantez {} kullanımını aktif ediyoruz.
profil = f"Öğrenci Adı: {isim} | Okulu: {okul} | Not Ortalaması: {not_ortalamasi}"
print(profil)

# Süslü parantezlerin içinde anlık matematiksel işlemler de yapabilirsiniz.
print(f"{isim} isimli öğrencinin notunu 5 puan artırırsak ortalaması {not_ortalamasi + 5} olur.")
```

## Adım 10: Kullanıcıdan Veri Alma (input fonksiyonu)

Programımız çalıştığında veriyi dışarıdan (klavyeden) almak istersek `input()` fonksiyonunu kullanırız. 
Kritik Kural: `input()` ile kullanıcıdan alınan her veri, içeriği sayı bile olsa, BİLGİSAYAR TARAFINDAN STRING (METİN) OLARAK OKUNUR.

```python
# Kullanıcıdan veri isteme süreci
kullanici_ismi = input("Lütfen isminizi giriniz: ")

# Yaş üzerinden matematiksel işlem yapacağımız için alınan string değeri hemen int() ile sayıya dönüştürüyoruz.
kullanici_yasi = input("Lütfen yaşınızı giriniz: ")
kullanici_yasi_sayi = int(kullanici_yasi)

print(f"Sisteme hoş geldin {kullanici_ismi}! Şu an {kullanici_yasi_sayi} yaşındasın.")
print(f"5 yıl sonra {kullanici_yasi_sayi + 5} yaşında olacaksın.")
```

## Adım 11: Dene ve Keşfet - String Parçalama (Slicing)

Sadece tek bir harfi almak yerine, metnin belirli bir parçasını kesip alabiliriz. Buna 'Slicing' (Dilimleme) adı verilir. 
Kullanımı: `metin[başlangıç_indeksi:bitiş_indeksi]` şeklindedir (Bitiş indeksi dahil edilmez).

Aşağıdaki kod bloğunda boş bırakılan `0:0` rakamlarını yönergelere göre değiştirin ve doğru sonucu bulmaya çalışın.

```python
uzun_kelime = "Karakteristik"

# İlk 4 harfi almak için:
ilk_dort_harf = uzun_kelime[0:4]
print("İlk 4 harf:", ilk_dort_harf)

# Görev: Siz de 'rakter' kısmını kesip almak için aşağıdaki 0:0 değerlerini doğru indekslerle değiştirin.
# İpucu: 'r' harfi 2. indekstedir. Bitiş indeksi kesime dahil edilmediği için bir fazlası yazılmalıdır.

kesilmis_parca = uzun_kelime[0:0] # BU SAYILARI DÜZELTİN (Örn: [2:8])
print("Kesilen Parça:", kesilmis_parca)
```
