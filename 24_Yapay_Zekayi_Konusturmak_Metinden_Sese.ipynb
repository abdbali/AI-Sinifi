# 24. Ders: Kendi Yapay Zekamızı Konuşturmak (Metinden Sese / TTS)

## Adım 1: Teorik Çerçeve ve Günlük Hayat Analojisi

Geçmişteki robot seslerini hatırlayın (Örneğin GPS navigasyon cihazlarındaki kesik kesik, duygusuz mekanik sesler). Bunlar eski nesil kelime birleştirme (concatenation) yöntemiyle yapılırdı. Bilgisayar kelimeleri stüdyoda daha önce kaydedilmiş parçalardan koparıp yan yana dizerdi, bu yüzden vurgular (duygu) tamamen kaybolurdu.

Bugün ise ChatGPT'nin sesli modu veya TikTok/Instagram videolarında duyduğunuz o pürüzsüz dış sesler (AI Voice) yepyeni bir teknoloji kullanır. Model kelimeleri koparıp yapıştırmaz; aldığı metnin içeriğine göre ses dalgalarını (frekansları) sıfırdan, pürüzsüzce 'üretir'. Cümle soru cümlesiyse sesin tonunu sonda inceltir. Bu sisteme Metinden Sese (Text-to-Speech / TTS) denir. Bu dersimizde sistemimizi konuşturacağız.

## Adım 2: Teknik Terimler ve Sözlük

- TTS (Text-To-Speech): Yazılı bir metni, insan sesine benzer bir şekilde (sentetik olarak) okuyan ses dosyalarına çeviren teknoloji.
- Speech Synthesis (Ses Sentezi): Makinenin dijital frekanslar oluşturarak hiç var olmamış bir sesi üretmesi.
- Voice Cloning (Ses Klonlama): Sisteme birinin (Örn: sizin) sesinden 10 saniyelik bir örnek verip, modelin yazılan tüm metinleri tamamen SİZİN sesinizle (veya ünlü birinin sesiyle) okumasıdır (Günümüzde dolandırıcılıkta kullanıldığı için çok tehlikeli ve regüle edilen bir konudur).
- gTTS (Google Text-to-Speech): Google'ın çeviri (Translate) altyapısını kullanan, öğrenmesi çok basit ve hafif bir Python TTS kütüphanesi.

## Adım 3: gTTS Kütüphanesini Kurma ve İçe Aktarma

Yapay zekayı çok zorlamamak ve hızlı sonuç almak için (Hugging Face'in devasa TTS modelleri yerine) başlangıç için `gTTS` kütüphanesini kullanacağız.

```python
!pip install gTTS
```

```python
from gtts import gTTS
from IPython.display import Audio # Colab'de sesi çalmak için

print("Ses sentezleme modülü yüklendi.")
```

## Adım 4: Türkçe Metni Sese Çevirme

Google'ın altyapısı sayesinde sistem Türkçe karakterleri (ş, ğ, ç) ve vurguları çok iyi anlar.

```python
okunacak_metin = "Merhaba! Ben senin bilgisayarında çalışan bir yapay zekayım. Bugün nasılsın?"

# gTTS komutu metni alır, lang="tr" ile Türkçe okunacağını belirtir.
ses_olusturucu = gTTS(text=okunacak_metin, lang="tr", slow=False)

# Üretilen sesi bilgisayara (sunucuya) mp3 olarak kaydetmeliyiz
dosya_adi = "yapay_zeka_sesi.mp3"
ses_olusturucu.save(dosya_adi)

print("Ses başarıyla sentezlendi ve mp3 olarak kaydedildi.")
```

## Adım 5: Üretilen Sesi Colab İçinde Oynatmak

Kaydettiğimiz mp3 dosyasını dinleyelim.

```python
# autoplay=True parametresi sayesinde, siz hücreyi çalıştırır çalıştırmaz ses otomatik çalmaya başlar.
display(Audio(dosya_adi, autoplay=True))
```

## Adım 6: Farklı Dillerde (Fransızca, Japonca vb.) Ses Çıktısı Alma

`lang` parametresini değiştirerek aynı metnin veya yabancı bir metnin o dilin aksanıyla nasıl okunduğunu görebilirsiniz.

```python
# Fransızca bir metin (fr)
fransizca_metin = "Bonjour mon ami, l'intelligence artificielle est magnifique."
fr_ses = gTTS(text=fransizca_metin, lang="fr")
fr_ses.save("fransizca.mp3")

print("Fransızca Aksanı:")
display(Audio("fransizca.mp3"))

# Not: Eğer Türkçe bir cümleyi lang="en" (İngilizce) okutursanız, 
# turist gibi komik bir aksanla okumaya çalışacaktır!
```
