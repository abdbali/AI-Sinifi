## Adım 2: Teknik Terimler ve Sözlük

- Algoritma: Bir problemin çözümü için izlenmesi gereken mantıksal, ardışık ve sonlu adımlar kümesi.
- Kod Hücresi (Code Cell): Bilgisayarın anlamlandırıp çalıştırabileceği, genellikle Python dilinde yazılmış metinleri barındıran bölümler.
- Metin Hücresi (Markdown Cell): Şu anda okumakta olduğunuz, kodların ne işe yaradığını açıklayan ve insan dilinde yazılan bilgilendirme bölümleri.
- CPU (Merkezi İşlem Birimi): Bilgisayarın genel amaçlı beyni. Temel matematiksel ve mantıksal işlemleri sırayla yapar.
- GPU (Grafik İşlem Birimi): Özellikle aynı anda binlerce paralel matematiksel hesaplama (matris çarpımı) yapabilen, yapay zeka eğitimlerinde süreci hızlandıran donanım.
- Çıktı (Output): Yazdığımız kodun bilgisayar tarafından işlendikten sonra bize döndürdüğü sonuç.

```python
# 'isim' adında bir değişken oluşturuyoruz ve içine bir metin (string) koyuyoruz.
isim = "Ahmet"

# 'yas' adında bir değişken oluşturuyoruz ve içine bir tam sayı (integer) koyuyoruz.
yas = 16

# Değişkenleri kullanarak ekrana daha dinamik çıktılar yazdırabiliriz.
print("Benim adım:")
print(isim)
print("Benim yaşım:")
print(yas)
```
