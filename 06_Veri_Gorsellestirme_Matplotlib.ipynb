# 6. Ders: Veri Görselleştirme - Sayılardan Çizgilere (Matplotlib)

## Adım 1: Teorik Çerçeve ve Günlük Hayat Analojisi

İnsan beyni karmaşık sayı tablolarını (örneğin 10.000 satırlık bir finans verisini) okumakta ve anlamlandırmakta çok zorlanır. Ancak aynı veriler yukarı doğru ivmelenen bir çizgi grafiği olarak karşımıza çıktığında, saniyeler içinde "Burada bir büyüme trendi var" kararını verebiliriz. 

Yapay zeka modellerini eğitirken, elimizdeki verinin dağılımını anlamak, modelin doğru öğrenip öğrenmediğini test etmek (Loss eğrisi) ve analizlerimizi diğer insanlara sunmak için Veri Görselleştirme tekniklerini kullanırız. Python ekosisteminde bu işin en temel ve en yaygın kullanılan kütüphanesi 'Matplotlib'dir. Bu dersimizde, elimizdeki sıkıcı verileri renkli ve anlamlı görsellere dönüştüreceğiz.

## Adım 2: Teknik Terimler ve Sözlük

- Matplotlib: Python'un 2 boyutlu grafik ve görselleştirme kütüphanesidir.
- Plot: Grafiği çizmek (render etmek) anlamına gelir.
- X Ekseni (X-Axis): Grafiğin yatay taban çizgisi. Genellikle bağımsız değişkenler (zaman, kategoriler) buraya konur.
- Y Ekseni (Y-Axis): Grafiğin dikey çizgisi. Genellikle ölçülen, değişen sayısal değerler buraya konur.
- Line Plot (Çizgi Grafiği): Noktaların birbirine çizgilerle bağlandığı, zaman içindeki değişimi gösteren grafik türü.
- Bar Chart (Sütun Grafiği): Kategoriler arasındaki farkları sütun uzunluklarıyla gösteren grafik türü.
- Scatter Plot (Dağılım Grafiği): Verilerin sadece birer nokta olarak uzaya bırakıldığı, aralarındaki ilişkinin (kümelenmenin) arandığı grafik türüdür. Yapay zekada çok önemlidir.

## Adım 3: Matplotlib Kütüphanesini İçe Aktarma

Grafik çizmek için Matplotlib'in `pyplot` isimli alt modülünü koda davet edeceğiz ve ona 'plt' kısaltmasını vereceğiz.

```python
# Grafik kütüphanesini içe aktarma
import matplotlib.pyplot as plt

print("Matplotlib başarıyla yüklendi ve grafik çizimine hazır.")
```

## Adım 4: Basit Bir Çizgi Grafiği (Line Plot) Çizmek

Çizgi grafikleri, iki farklı veri listesinin (X ve Y koordinatları) eşleştirilmesiyle oluşur. Örneğin günlere göre sıcaklık değişimi gibi.

```python
gunler = ["Pzt", "Sal", "Çar", "Per", "Cum"]
sicakliklar = [15, 17, 14, 20, 22]

# plt.plot(X_Ekseni_Verisi, Y_Ekseni_Verisi)
plt.plot(gunler, sicakliklar)

# Çizimi ekranda göstermek için show() metodu çağrılmalıdır.
plt.show()
```

## Adım 5: Grafiği Özelleştirmek ve Kimlik Kazandırmak

Yukarıdaki grafik çalıştı ama üzerinde hiçbir bilgi yok. İyi bir veri bilimci grafiğe başlık, eksen isimleri ve anlamlı renkler atamalıdır.

```python
# Renk (color), çizgi stili (linestyle) ve nokta belirteci (marker) özelliklerini ekliyoruz.
plt.plot(gunler, sicakliklar, color="red", linestyle="--", marker="o")

# Grafiğe Ana Başlık ekleme
plt.title("Haftalık Hava Sıcaklığı Değişimi")

# Eksen İsimleri (Label) ekleme
plt.xlabel("Haftanın Günleri")
plt.ylabel("Sıcaklık (Santigrat)")

# Arka plana ızgara (grid) çizgileri ekleme
plt.grid(True)

plt.show()
```

## Adım 6: Sütun Grafiği (Bar Chart)

Eğer zaman içindeki değişimi değil de, farklı kategoriler arasındaki farkları (Örn: Hangi programlama dili daha popüler?) göstermek istiyorsak Bar Chart kullanırız.

```python
diller = ["Python", "Java", "C++", "JavaScript"]
kullanici_sayisi = [85, 45, 30, 90]

# .plot yerine .bar kullanıyoruz
plt.bar(diller, kullanici_sayisi, color="teal")

plt.title("Programlama Dilleri Popülerlik Kıyaslaması")
plt.xlabel("Diller")
plt.ylabel("Kullanıcı Skoru")

plt.show()
```

## Adım 7: Dağılım Grafiği (Scatter Plot) ve Yapay Zekadaki Yeri

Makine öğrenmesinde verilerin uzayda nasıl dağıldığını (örneğin iki ayrı grubun birbirinden net ayrılıp ayrılamadığını) görmek için Scatter (Dağılım) grafikleri kullanılır. Sadece noktalar basılır, çizgiler birleştirilmez.

```python
# Öğrencilerin çalışma saatleri ve aldıkları sınav notları
calisma_saatleri = [1, 2, 2.5, 3, 4, 4.5, 5, 6, 7]
sinav_notlari = [40, 45, 50, 60, 65, 75, 80, 85, 95]

# .scatter metodu noktalar üretir
plt.scatter(calisma_saatleri, sinav_notlari, color="purple")

plt.title("Çalışma Saati ile Sınav Notu Arasındaki İlişki")
plt.xlabel("Çalışma Süresi (Saat)")
plt.ylabel("Sınav Notu")

plt.show()
# Grafiği incelediğinizde çalışma saati arttıkça notun da doğrusallıkla arttığını görebilirsiniz.
```

## Adım 8: Pandas ile Matplotlib Birlikteliği

Önceki dersimizde öğrendiğimiz Pandas veri çerçeveleri (DataFrame) üzerinden verileri çekip doğrudan grafik çizebiliriz. Bu iki kütüphane birbiriyle uyumlu çalışır.

```python
import pandas as pd

veriler = {
    "Şehirler": ["İstanbul", "Ankara", "İzmir", "Bursa"],
    "Nüfus (Milyon)": [16.0, 5.7, 4.4, 3.1]
}

df = pd.DataFrame(veriler)

# DataFrame'den sütunları çekerek doğrudan plt içerisine veriyoruz.
plt.bar(df["Şehirler"], df["Nüfus (Milyon)"], color="orange")
plt.title("Büyük Şehirlerin Nüfus Dağılımı")
plt.show()
```

## Adım 9: Dene ve Keşfet - Grafik Türünü Değiştirme

Aşağıdaki kodda iki farklı yılın aylık kazançları verilmiştir. Bu grafik şu an çizgi (plot) grafiği olarak çizilmektedir. Kodu değiştirerek bunu bir 'Scatter' (Dağılım) grafiğine dönüştürün ve inceleyin.

```python
aylar = [1, 2, 3, 4, 5]
kazanc = [2000, 2200, 2500, 2100, 2800]

# Bu satırı değiştirin (plt.plot yerine ne gelmeli?)
plt.plot(aylar, kazanc, marker="x")

plt.show()
```
