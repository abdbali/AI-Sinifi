## Adım 1: Teorik Çerçeve ve Günlük Hayat Analojisi

Bir bilgisayarın insandan en büyük farkı asla yorulmaması ve sıkılmamasıdır. Bir insana 'Ekrana 10.000 kere Merhaba yaz' derseniz bir süre sonra hata yapmaya başlar veya işi bırakır. Bilgisayarlar ise bunu milisaniyeler içinde kusursuzca yapar. Programlamada, aynı işlemin defalarca tekrar edilmesini sağlayan bu yapıya 'Döngü' (Loop) denir.

Ayrıca kod yazarken bazı işlemleri sık sık tekrar ederiz (Örneğin sürekli Vücut Kitle İndeksi hesaplamak gibi). ## Adım 2: Teknik Terimler ve Sözlük

- Loop (Döngü): Belirli bir kod bloğunu ardışık olarak tekrar tekrar çalıştıran yapıdır.
- For Döngüsü: İşlemin kaç kere tekrar edileceği önceden belli olan durumlarda kullanılır.
- While Döngüsü: İşlemin tekrar sayısı belli olmayan, sadece belirli bir koşul doğru (True) olduğu sürece devam etmesi istenen durumlarda kullanılır.
- Iteration (İterasyon): Döngünün her bir adımına (her bir turuna) verilen isimdir.
- List (Liste): İçerisinde birden fazla veriyi aynı anda tutabilen, köşeli parantez '[]' ile tanımlanan veri tipidir.
- Function (Fonksiyon): Belirli bir görevi yerine getiren, ismi çağrıldığında çalışan paketlenmiş kod bloğudur.
- Def (Define): Python'da yeni bir fonksiyon tanımlamak için kullanılan anahtar kelimedir.

## Adım 3: Listeler (List) Veri Tipine Hızlı Bakış

Döngülere geçmeden önce, içinde birden fazla veri barındıran Listeleri öğrenmeliyiz. Listeler yapay zeka modellerine veri yığınlarını beslerken en çok kullanacağımız araçlardır.

```python
# İçerisinde 4 farklı metin barındıran bir liste
ogrenciler = ["Ali", "Ayşe", "Mehmet", "Fatma"]
print("Öğrenci Listesi:", ogrenciler)

# Listeler de stringler gibi 0. indeksten başlar
print("Listenin ilk elemanı:", ogrenciler[0])
print("Listenin üçüncü elemanı:", ogrenciler[2])

# Listeye yeni eleman eklemek için .append() metodu kullanılır
ogrenciler.append("Zeynep")
print("Yeni Öğrenci Eklendikten Sonra:", ogrenciler)
```

## Adım 4: For Döngüsü (Belirli Sayıda Tekrar)

Eğer bir listenin içindeki her bir elemana tek tek gidip onlar üzerinde işlem yapmak istiyorsanız 'for' döngüsü kullanılır.

```python
notlar = [85, 90, 78, 92, 60]

# Okunuşu: 'notlar' listesinin içindeki her bir elemanı sırayla 'tekil_not' değişkenine ata ve bloğu çalıştır.
for tekil_not in notlar:
    print(f"Sistemdeki not okundu: {tekil_not}")
    # Döngü içinde if-else kararları da verebiliriz
    if tekil_not >= 80:
        print("-> Bu not oldukça başarılı!")
        
print("\nTüm notların okunma işlemi bitti.")
```

## Adım 5: Range Fonksiyonu ile Tekrar Sayısı Belirleme

Elimizde bir liste yoksa ancak sadece kodun 5 defa veya 100 defa çalışmasını istiyorsak `range()` fonksiyonunu kullanırız.

```python
# range(5) komutu bilgisayara 0'dan başlayarak 4'e kadar (5 dahil değil) saymasını söyler.
for adim in range(5):
    print(f"{adim}. iterasyon (tur) çalışıyor...")
    
# Belirli bir aralık da verebiliriz (Örn: 10'dan 15'e kadar)
print("\nYeni bir döngü başlıyor:")
for sayi in range(10, 15):
    print("Sayı:", sayi)
```

## Adım 6: While Döngüsü (Koşula Bağlı Tekrar)

Bazen bir işlemin kaç kere tekrar edeceğini bilemeyiz. Örneğin kullanıcı doğru şifreyi girene kadar ona şifre sormak isteriz. Bu durumda `while` (olduğu sürece) döngüsü kullanılır.

```python
enerji_seviyesi = 3

# Enerji seviyesi 0'dan büyük OLDUĞU SÜRECE bu bloğu tekrarla.
while enerji_seviyesi > 0:
    print(f"Robot çalışıyor... Kalan enerji: {enerji_seviyesi}")
    # Her turda enerjiyi 1 azaltmalıyız, aksi halde sonsuz döngüye gireriz!
    enerji_seviyesi = enerji_seviyesi - 1

print("Enerji bitti. Robot durdu.")
```

## Adım 7: Sonsuz Döngüler ve Break Komutu

Bazen bilinçli olarak sonsuz bir döngü (`while True`) başlatırız ve içeride bir koşul gerçekleştiğinde döngüyü zorla kırmak için `break` (kır) komutunu kullanırız.

```python
sayac = 1

while True: # Sonsuz döngü!
    print(f"Sayım: {sayac}")
    sayac = sayac + 1
    
    if sayac == 6:
        print("Tehlike sınırı aşıldı, döngü zorla kırılıyor!")
        break # Bu komut döngüyü anında sonlandırır.

print("Döngüden güvenle çıkıldı.")
```

## Adım 8: Fonksiyon Tanımlamaya Giriş (def)

Şimdiye kadar hep Python'un hazır fonksiyonlarını (`print()`, `input()`, `len()`) kullandık. Artık kendi fonksiyonlarımızı yaratma zamanı. `def` kelimesi 'Define' (Tanımla) kelimesinin kısaltmasıdır.

```python
# 'karsilama_yap' adında bir fonksiyon paketliyoruz.
def karsilama_yap():
    print("Sisteme hoş geldiniz.")
    print("Tüm yapay zeka birimleri aktif edildi.")

# Fonksiyonlar tanımlandıklarında çalışmazlar. Sadece isimleriyle çağrıldıklarında çalışırlar.
print("Program başladı...")
karsilama_yap() # Fonksiyonu çağırıyoruz
karsilama_yap() # İkinci kez çağırıyoruz
```

## Adım 9: Parametre (Girdi) Alan Fonksiyonlar

Fonksiyonlar dışarıdan veri alıp, o veriye özel işlem yapabilirler. Buna parametre vermek denir.

```python
# İçerisine 'isim' adında bir parametre (veri) bekleyen fonksiyon
def ozel_karsilama(isim):
    print(f"Merhaba {isim}, yetkilendirme onaylandı.")

ozel_karsilama("Ali")
ozel_karsilama("Ayşe")

# Birden fazla parametre de alabilir
def alan_hesapla(kisa_kenar, uzun_kenar):
    alan = kisa_kenar * uzun_kenar
    print(f"Kenarları {kisa_kenar} ve {uzun_kenar} olan dikdörtgenin alanı: {alan}")

alan_hesapla(5, 10)
alan_hesapla(7, 3)
```

## Adım 10: Geri Değer Döndüren (Return) Fonksiyonlar

Çoğu zaman bir fonksiyonun sadece ekrana yazı yazdırmasını istemeyiz, bir hesaplama yapıp sonucu bize (koda) geri vermesini (return) bekleriz.

```python
def kup_hesapla(sayi):
    sonuc = sayi * sayi * sayi
    # print kullanmak yerine veriyi sisteme geri döndürüyoruz
    return sonuc

# Fonksiyondan geri dönen değeri yeni bir değişkene atayabiliriz
hesaplanan_deger = kup_hesapla(4)
print("4'ün küpü:", hesaplanan_deger)

print("3'ün küpü:", kup_hesapla(3))
```
