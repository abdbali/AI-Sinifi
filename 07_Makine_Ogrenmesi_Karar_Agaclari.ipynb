# 7. Ders: Makine Öğrenmesi Mantığı - Karar Ağaçları ve Sınıflandırma

## Adım 1: Teorik Çerçeve ve Günlük Hayat Analojisi

Şu ana kadar if-else yapılarıyla bilgisayara kararları BİZ verdirdik. 'Eğer hız 120'den büyükse ceza kes' dedik. Bu geleneksel programlamadır (Kural + Veri = Sonuç). Ancak, spam e-postaları veya yüz tanımasını if-else ile yazamazsınız çünkü kurallar çok karmaşıktır.

İşte burada 'Makine Öğrenmesi' (Machine Learning) devreye girer. Makine öğrenmesinde bilgisayara Kuralları BİZ VERMEYİZ. Sadece Verileri ve Sonuçları (Cevapları) veririz. Bilgisayar bu verilere bakarak 'Kuralları kendisi öğrenir'. Buna 'Eğitim' (Training) denir. Bu dersimizde en anlaşılır yapay zeka algoritmalarından biri olan Karar Ağaçlarını (Decision Trees) kullanarak, kendi kurallarını bulan bir teşhis modeli yazacağız.

## Adım 2: Teknik Terimler ve Sözlük

- Scikit-Learn (sklearn): Python'un makine öğrenmesi için geliştirilmiş en temel, ücretsiz matematik ve algoritma kütüphanesidir.
- Feature (Öznitelik - X): Modelin tahmin yapmak için kullanacağı ipuçları veya özellikler (Örn: Hastanın yaşı, ateşi).
- Label (Etiket - y): Modelin bulmaya çalıştığı sonuç veya hedef (Örn: Hasta mı? Değil mi?).
- Training (Eğitim - fit): Modelin geçmiş verilere bakarak kural (örüntü) çıkarması işlemidir.
- Predict (Tahmin Etmek): Eğitilmiş modelin daha önce hiç görmediği yeni bir veri üzerinde fikir yürütmesidir.
- Decision Tree (Karar Ağacı): Veriyi yukarıdan aşağıya if-else benzeri dallara bölerek sınıflandıran bir algoritma türüdür.

## Adım 3: Scikit-Learn ve Pandas Kütüphanelerini İçe Aktarma

Önceki derslerde öğrendiğimiz Pandas'ı veriyi düzenlemek için, Scikit-Learn'ü ise modelimizi oluşturmak için kullanacağız.

```python
import pandas as pd

# sklearn kütüphanesinden Karar Ağacı (DecisionTreeClassifier) aracını çağırıyoruz.
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

print("Makine Öğrenmesi araçları başarıyla koda dahil edildi.")
```

## Adım 4: Yapay Zeka İçin Eğitim Verisi Oluşturma

Kurgusal bir senaryo oluşturalım. Hastanede çalışan bir doktorsunuz. Hastaların ateşini (derece) ve oksijen seviyelerini ölçüyorsunuz ve sonuçta hastaya 'Soğuk Algınlığı' (1) veya 'Sağlıklı' (0) teşhisi koyuyorsunuz.

```python
# Gemişteki hastaların kayıtları
hasta_kayitlari = {
    "Ates": [36.5, 38.5, 39.0, 36.6, 37.0, 38.0, 36.4],
    "Oksijen": [98, 92, 90, 99, 97, 94, 98],
    "Hastalik_Durumu": [0, 1, 1, 0, 0, 1, 0] # 0: Sağlıklı, 1: Hasta
}

df = pd.DataFrame(hasta_kayitlari)
print("Eğitim İçin Kullanılacak Veri Seti (Geçmiş Kayıtlar):")
print(df)
```

## Adım 5: Veriyi X (Özellikler) ve y (Hedef) Olarak İkiye Bölmek

Yapay zeka modellerinin kuralı şudur: Ona çalışacağı ipuçlarını (X) ve bulması gereken cevapları (y) ayrı ayrı vermelisiniz.

```python
# X (Büyük harf yazılır): Sadece Ateş ve Oksijen sütunlarını alıyoruz. Sonucu (Hastalık_Durumu) siliyoruz.
X = df.drop(columns=["Hastalik_Durumu"])

# y (Küçük harf yazılır): Sadece makinenin öğrenmesini istediğimiz hedeftir.
y = df["Hastalik_Durumu"]

print("Öğrenilecek İpuçları (X):\n", X.head(2))
print("\nHedef Cevaplar (y):\n", y.head(2))
```

## Adım 6: Makine Öğrenmesi Modelini Tanımlamak

Tıpkı boş bir beyin (veya yeni doğmuş bir bebek) oluşturur gibi algoritmamızı kodluyoruz.

```python
# Karar ağacı sınıflandırıcısını oluşturuyoruz.
model = DecisionTreeClassifier()

print("Model oluşturuldu, ancak henüz hiçbir şey öğrenmedi (eğitilmedi).")
```

## Adım 7: Modeli Eğitmek (Fit)

Şimdi bu boş beyne, geçmişteki `X` (ipuçları) ve `y` (cevaplar) verilerini vereceğiz. Model saniyeler içinde tüm istatistiği tarayacak ve kendi matematiksel if-else kurallarını çıkaracak.

```python
# .fit() komutu makinenin öğrenme (eğitim) işlemini başlatır.
model.fit(X, y)

print("Tebrikler! Yapay Zeka modeli verileri inceledi ve kuralları öğrendi.")
```

## Adım 8: Yeni Veri Üzerinde Tahmin Yapmak (Predict)

Şimdi hastaneye yeni biri geldi. Ateşi 38.2, Oksijeni ise 93. Makineye bu yeni ipuçlarını verelim ve bakalım bu kişiye ne teşhis koyacak?

```python
# Yeni hastanın verisini Pandas DataFrame olarak vermeliyiz (Model X'i öyle öğrendiği için)
yeni_hasta = pd.DataFrame({"Ates": [38.2], "Oksijen": [93]})

# .predict() metodu ile tahmin yapıyoruz.
tahmin = model.predict(yeni_hasta)

if tahmin[0] == 1:
    print("Yapay Zeka Teşhisi: HASTA (1)")
else:
    print("Yapay Zeka Teşhisi: SAĞLIKLI (0)")
```

## Adım 9: Karar Ağacını Görselleştirmek

Makine öğrenmesinin en güzel yanı, Karar Ağaçlarının arka planda hangi kuralı bulduğunu bizim de görebilmemizdir. Acaba bizi neye göre Hasta veya Sağlıklı olarak ayırdı?

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
# plot_tree fonksiyonu ağacın dallarını çizer.
plot_tree(model, feature_names=["Ates", "Oksijen"], class_names=["Sağlıklı", "Hasta"], filled=True)
plt.show()

# Grafiğin en üstündeki kutuya (Root/Kök) bakın. 
# Model büyük ihtimalle "Ateş <= 37.5" ise sağlıklıdır gibi bir kuralı KENDİLİĞİNDEN bulmuştur!
```

## Adım 10: Dene ve Keşfet - Farklı Hastalar

Aşağıdaki kod bloğunda, hastaneye aynı anda başvuran 3 farklı kişinin ateşi ve oksijen seviyeleri liste olarak verilmiştir. Kodu çalıştırarak modelin üçüne birden aynı anda nasıl teşhis koyduğunu inceleyin.

```python
# 3 Farklı Yeni Hasta
acilden_gelenler = pd.DataFrame({
    "Ates": [36.2, 39.5, 37.1],
    "Oksijen": [99, 88, 96]
})

# Tahminleri alalım (Sonuç 3 elemanlı bir liste olacak)
toplu_tahminler = model.predict(acilden_gelenler)

print("Sırasıyla 3 hastanın durumları (0: Sağlam, 1: Hasta):")
print(toplu_tahminler)
```
