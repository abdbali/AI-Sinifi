# 3. Ders: Karar Mekanizmaları ve Kontrol Yapıları (If, Elif, Else)

## Adım 1: Teorik Çerçeve ve Günlük Hayat Analojisi

Hayatımız boyunca her gün yüzlerce karar veririz: 'Eğer hava yağmurluysa şemsiye al, değilse alma.', 'Eğer saat 08:00 olduysa uyan, aksi halde uyumaya devam et.' Bilgisayarlar da tıpkı bu şekilde, önüne çıkan veri ve durumlara göre belirli yolları seçmek zorundadır. Programlama dilinde buna 'Koşullu İfadeler' (Conditional Statements) denir.

Geçen haftaki dersimizde Boolean (True/False) veri tipini öğrenmiştik. Bir yapay zeka modelinin veya herhangi bir uygulamanın akıllı davranabilmesi için, doğru (True) olan senaryolarda ayrı bir kod bloğunu, yanlış (False) olan senaryolarda ayrı bir kod bloğunu çalıştırması gerekir. Bu derste, bilgisayarımıza kendi analizlerini yaparak 'Eğer öyleyse şunu yap, yoksa bunu yap' diyebilmeyi öğreneceğiz.

## Adım 2: Teknik Terimler ve Sözlük

- If (Eğer): Belirtilen bir koşulun 'True' (Doğru) olması durumunda çalışacak kod bloğunu başlatır.
- Else (Değilse / Aksi Halde): Yukarıdaki hiçbir 'if' veya 'elif' koşulu sağlanmadığında varsayılan olarak çalışacak son çıkış kapısıdır. Bir koşula bağlı değildir.
- Elif (Else If / Değilse Eğer): İlk 'if' koşulu sağlanmadıysa kontrol edilecek ikinci, üçüncü veya dördüncü alternatif koşulları belirtir.
- Condition (Koşul): Sonucu her zaman True veya False olan test durumudur (Örn: kullanici_yasi > 18).
- Indentation (Girintileme): Python'da kod bloklarının birbirinden ayrılması için satır başlarına bırakılan boşluklardır (Genellikle klavyedeki TAB tuşu ile yapılır).

## Adım 3: Mantıksal Karşılaştırma Operatörleri

Karar mekanizmalarını kurabilmek için bilgisayarın iki değeri birbiriyle nasıl kıyaslayacağını bilmesi gerekir. Python'da en sık kullanılan 6 temel karşılaştırma operatörü vardır ve bunların ürettiği sonuç her zaman bir Boolean değeridir.

```python
A = 10
B = 25

print("Eşitlik Kontrolü (==):", A == B)         # İki değer birbirine eşit mi?
print("Eşitsizlik Kontrolü (!=):", A != B)      # İki değer birbirinden farklı mı?
print("Büyüklük Kontrolü (>):", A > B)          # A, B'den büyük mü?
print("Küçüklük Kontrolü (<):", A < B)          # A, B'den küçük mü?
print("Büyük Eşit Kontrolü (>=):", A >= 10)     # A, 10'a eşit veya ondan büyük mü?
print("Küçük Eşit Kontrolü (<=):", B <= 20)     # B, 20'ye eşit veya ondan küçük mü?
```

## Adım 4: Mantıksal Bağlaçlar (and, or, not)

Bazen tek bir karar vermek için birden fazla şartın aynı anda değerlendirilmesi gerekir. Örneğin bir arabayı sürebilmek için hem ehliyetinizin olması 've' (and) hem de arabanın anahtarının olması gerekir.

```python
sinavdan_gecti_mi = True
devamsizlik_hakki_var_mi = False

# 'and' bağlacı: Her iki taraf da True olursa genel sonuç True olur. Aksi halde False olur.
sinifi_gecme_durumu = sinavdan_gecti_mi and devamsizlik_hakki_var_mi
print("Öğrenci sınıfı geçti mi? (AND):", sinifi_gecme_durumu)

kredi_karti_var = False
nakit_para_var = True

# 'or' bağlacı: Taraflardan sadece birinin True olması, genel sonucun True olması için yeterlidir.
alisveris_yapabilir_mi = kredi_karti_var or nakit_para_var
print("Müşteri alışveriş yapabilir mi? (OR):", alisveris_yapabilir_mi)

# 'not' bağlacı: Durumu tersine çevirir. True ise False, False ise True yapar.
sistem_kapali_mi = False
print("Sisteme giriş yapılabilir mi? (NOT):", not sistem_kapali_mi)
```

## Adım 5: If Bloğunun Temel Yapısı ve Girintileme (Indentation)

Koşullu ifadeleri yazarken Python'un en önemli dil bilgisi kuralı olan 'girintileme' ile tanışacağız. `if` komutunun sonuna iki nokta (`:`) konulur ve bir alt satıra geçildiğinde kod hafif sağdan (4 boşluk veya bir TAB) başlar. Bu içerideki kodun, sadece koşul doğruysa çalışacağını belirtir.

```python
hava_durumu = "Yağmurlu"

# Eğer hava_durumu 'Yağmurlu' kelimesine tam olarak eşitse altındaki bloğu çalıştır.
if hava_durumu == "Yağmurlu":
    # Burası bir kod bloğudur. Girintili olduğu için if yapısına aittir.
    print("Dışarı çıkarken mutlaka şemsiyeni yanına al.")
    print("Yağmurluk giymeyi de unutma.")

# Bu satır girintili olmadığı için if bloğunun DIŞINDADIR. 
# Yukarıdaki şart sağlansa da sağlanmasa da bu kod her halükarda çalışır.
print("Hava durumu kontrolü tamamlandı.")
```

## Adım 6: If-Else Yapısı (İki Seçenekli Durumlar)

Sadece koşul sağlandığında değil, sağlanmadığında da farklı bir işlem yapılmasını istiyorsak `else` yapısını kullanırız. Tıpkı madeni paranın yazı veya tura gelmesi gibi kesin bir ayrım sunar.

```python
girilen_parola = "gizliSifre123"
sistem_parolasi = "yapayzeka2026"

if girilen_parola == sistem_parolasi:
    print("Sisteme başarıyla giriş yapıldı! Yüksek gizlilikteki veriler yükleniyor...")
else:
    print("Hatalı parola girdiniz! Lütfen tekrar deneyin.")
    
# Görev: 1. satırdaki 'girilen_parola' değişkenini 'yapayzeka2026' olarak değiştirip kodu tekrar çalıştırarak sonucu gözlemleyin.
```

## Adım 7: If-Elif-Else Yapısı (Çok Seçenekli Durumlar)

Gerçek hayatta seçenekler her zaman sadece iki tane (siyah veya beyaz) değildir. Gri tonlarının veya tamamen farklı seçeneklerin olduğu durumlarda `elif` bloğunu araya ekleriz. Tek bir `if` bloğu içinde istediğiniz kadar `elif` kullanabilirsiniz.

```python
sinav_notu = 75

# Bilgisayar kodu yukarıdan aşağıya doğru sırayla okur.
# Hangi şart İLK olarak sağlanırsa, sadece o bloğu çalıştırır ve sistemi terk eder.
if sinav_notu >= 90:
    print("Harf Notunuz: A (Mükemmel)")
elif sinav_notu >= 80:
    print("Harf Notunuz: B (Çok İyi)")
elif sinav_notu >= 70:
    print("Harf Notunuz: C (İyi)")
elif sinav_notu >= 60:
    print("Harf Notunuz: D (Geçer)")
else:
    print("Harf Notunuz: F (Kaldınız)")
```

## Adım 8: İç İçe (Nested) If Yapıları

Bazen bir koşulun içinde, sadece o koşul sağlandığında kontrol edilecek spesifik alt koşullar olabilir. Bu durumda kodlar daha da içe doğru girintilenerek yazılır.

```python
uyelik_aktif_mi = True
uyelik_tipi = "Premium"

if uyelik_aktif_mi == True:
    print("Sisteme başarıyla giriş yaptınız.")
    
    # Bu iç 'if' bloğu, sadece üyelik aktifse çalışacak şekilde girintilenmiştir.
    if uyelik_tipi == "Premium":
        print("Tüm yapay zeka özelliklerine sınırsız erişiminiz var.")
    else:
        print("Sınırlı erişim. Bazı özellikler için hesabınızı yükseltin.")

# Dışarıdaki else bloğu, sadece uyelik_aktif_mi değişkeni False ise çalışır.
else:
    print("Üyeliğiniz bulunmamaktadır. Lütfen kayıt olun.")
```

## Adım 9: Dene ve Keşfet - Mantık Hataları (Bug Analizi)

Bilgisayarlar kendi kendine mantık yürütmez, sadece sizin yazdığınız sıralamayı tavizsiz bir şekilde takip eder. Aşağıdaki kodda ciddi bir mantıksal sıralama hatası (bug) var. Notu 95 olan bir öğrencinin 'A' alması gerekirken farklı bir not alacaktır.

Neden böyle olduğunu düşünün ve if-elif sıralamasını mantıklı bir şekilde değiştirerek bu sistemi düzeltin.

```python
ogrencinin_notu = 95

# DİKKAT: Hatalı sıralama örneğidir. Lütfen düzeltiniz.
if ogrencinin_notu >= 50:
    print("Dersten Geçtiniz!")
elif ogrencinin_notu >= 70:
    print("İyi bir notla geçtiniz.")
elif ogrencinin_notu >= 90:
    print("Mükemmel bir notla geçtiniz!")
else:
    print("Dersten Kaldınız.")
    
# İpucu: Bilgisayar ilk doğru bulduğu koşulu çalıştırır. 95 rakamı aynı zamanda 50'den büyük olduğu için ilk if bloğuna girer ve işi bitirir.
# Çözüm olarak, en zorlu (en yüksek) koşul her zaman en üstte kontrol edilmelidir.
```

## Adım 10: Dene ve Keşfet - Çoklu Koşulları Birleştirme

Aşağıdaki banka kredi sistemini inceleyin. Koşulların parantezler içerisine alınarak `and` bağlacı ile birleştirilmesine dikkat edin.

```python
# Dinamik deneme için aşağıdaki değişkenlerin değerlerini değiştirebilirsiniz.
age = 25
salary = 15000
kredi_notu_yuksek_mi = True

# Şart: Yaşı 18'den büyük OLMALI VE maaşı 10000 üzerinde OLMALI VE kredi notu yüksek OLMALI.
if (age >= 18) and (salary >= 10000) and (kredi_notu_yuksek_mi == True):
    print("Kredi başvurunuz ONAYLANDI.")
else:
    print("Kredi başvurunuz REDDEDİLDİ. Şartları sağlamıyorsunuz.")
```
