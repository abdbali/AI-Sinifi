İnsanlar dünyaya baktığında renkleri, şekilleri, kedileri, ağaçları görür. Ancak bir bilgisayar fotoğraf makinesinden gelen veriye baktığında sadece uçsuz bucaksız bir sayılar denizi (sayı tablosu) görür.

Ekrandaki her bir görüntü, aslında 'Piksel' adı verilen milyonlarca küçük kareden oluşur. Bilgisayar için 'kırmızı' bir renk yoktur, sadece '255, 0, 0' sayıları vardır. Bu dersimizde dünyanın en meşhur görüntü işleme kütüphanesi olan OpenCV'yi kullanarak, bilgisayarın dünyayı nasıl gördüğünü anlayacağız.

## Adım 2: Teknik Terimler ve Sözlük

- Computer Vision (CV - Bilgisayarlı Görme): Bilgisayarların dijital resimlerden veya videolardan yüksek seviyeli anlam çıkarma işlemi.
- Pixel (Piksel): Ekranda renk gösterebilen en küçük nokta birimi (Picture Element kısaltmasıdır).
- Resolution (Çözünürlük): Ekranda veya fotoğrafta bulunan toplam piksel sayısını (Genişlik x Yükseklik) ifade eder.
- RGB (Red, Green, Blue): Tüm dijital renklerin Kırmızı, Yeşil ve Mavi ışıkların karışımıyla oluşturulduğu renk kodlama sistemi.
- Grayscale (Gri Tonlama): Renklerin (RGB) çöpe atılıp görüntünün sadece siyah-beyaz tonlarıyla ifade edilmesi.
- OpenCV (Open Source Computer Vision Library): Görüntü ve video işleme için kullanılan efsanevi C++ ve Python kütüphanesi.

## Adım 3: OpenCV ve Matplotlib Kütüphanelerini İçe Aktarma

OpenCV görüntüleri sayılara (matrislere) çevirir, daha önce öğrendiğimiz Matplotlib ise bu matrisleri bizim için ekranda çizdirir.

```python
import cv2 # OpenCV'nin Python'daki kısa adıdır
import matplotlib.pyplot as plt
import urllib.request # İnternetten fotoğraf indirmek için
import numpy as np # Matris (Sayı tablosu) işlemleri için

print("Görüntü İşleme araçları başarıyla yüklendi.")
```

## Adım 4: İnternetten Örnek Bir Fotoğraf İndirme

Test etmek için bir manzara veya hayvan fotoğrafı indirelim ve OpenCV ile bilgisayarın hafızasına alalım.

```python
# Örnek bir köpek fotoğrafı url'si
url = "https://upload.wikimedia.org/wikipedia/commons/thumb/3/3a/Cat03.jpg/800px-Cat03.jpg"
dosya_adi = "kedi.jpg"

# Fotoğrafı internetten indirip Colab içine kaydediyoruz
urllib.request.urlretrieve(url, dosya_adi)

# OpenCV'nin imread (Image Read - Görüntü Oku) komutuyla dosyayı sayılara çeviriyoruz
resim_matrisi = cv2.imread(dosya_adi)

print("Kedi fotoğrafı başarıyla okundu ve sayılara dönüştürüldü!")
```

## Adım 5: Görüntüyü Ekrana Çizdirme (Ve Meşhur BGR Sorunu)

Bilgisayarın hafızasındaki o sayıları tekrar resim olarak görelim. Ancak çok garip bir renkle karşılaşacağız!

```python
# Matplotlib ile resmi ekrana çizdirelim
plt.imshow(resim_matrisi)
plt.axis("off") # Eksen çizgilerini kapat
plt.show()

# FARK ETTİNİZ Mİ?
# Kedinin renkleri çok tuhaf (Mavimsi). Bunun sebebi, tüm dünya renkleri RGB (Kırmızı, Yeşil, Mavi)
# sırasıyla okurken, OpenCV'nin bunu tersten BGR (Mavi, Yeşil, Kırmızı) sırasıyla okumasıdır.
```

## Adım 6: BGR'dan RGB'ye Renk Düzeltme

Yapay zeka veya gösterim öncesinde mutlaka OpenCV'nin BGR formatını standart RGB formatına çevirmeliyiz.

```python
# cvtColor (Convert Color) komutu ile BGR'yi RGB'ye çeviriyoruz
duzgun_resim = cv2.cvtColor(resim_matrisi, cv2.COLOR_BGR2RGB)

plt.imshow(duzgun_resim)
plt.axis("off")
plt.title("Düzeltilmiş Gerçek Renkler")
plt.show()
```

## Adım 7: Görüntünün Matematiksel Boyutları (Shape)

Python için bu fotoğraf aslında 3 boyutlu devasa bir küptür (Yükseklik x Genişlik x Renk Kanalları).

```python
# .shape komutu matrisin boyutlarını verir
boyutlar = duzgun_resim.shape

print(f"Fotoğrafın Yüksekliği (Satır Sayısı): {boyutlar[0]} piksel")
print(f"Fotoğrafın Genişliği (Sütun Sayısı): {boyutlar[1]} piksel")
print(f"Renk Katmanları (Kanalları): {boyutlar[2]} (Red, Green, Blue)")

toplam_piksel = boyutlar[0] * boyutlar[1]
print(f"\nToplam {toplam_piksel} adet piksel, her biri 3 ayrı sayıdan oluştuğu için")
print(f"bilgisayar aslında şu an {toplam_piksel * 3} adet sayıyı aklında tutuyor!")
```

## Adım 8: Görüntüyü Siyah Beyaza Çevirme Neden Önemlidir?

Bilgisayar yukarıda gördüğümüz gibi milyonlarca sayıyla başa çıkmak zorundadır. Yapay zeka kedinin yüzünü bulacaksa, kedinin sarı veya gri olmasının bir önemi yoktur (Işık ve gölge yeterlidir). Bu yüzden performansı (hızı) tam 3 katına çıkarmak için fotoğrafları Gri Tonlamaya (Grayscale) çeviririz.

```python
# BGR'dan GRAY (Gri) formata çeviriyoruz
gri_resim = cv2.cvtColor(resim_matrisi, cv2.COLOR_BGR2GRAY)

print("Gri resmin boyutu:", gri_resim.shape)
# Çıktıda sadece (Yükseklik, Genişlik) göreceğiz. Sonundaki 3 rakamı (Renkler) yok oldu! Veri üçte birine düştü.

# Matplotlib gri çizerken özel bir renk haritası (cmap) ister
plt.imshow(gri_resim, cmap='gray')
plt.axis("off")
plt.title("Yapay Zeka İçin Optimize Edilmiş Siyah-Beyaz Görüntü")
plt.show()
```

## Adım 9: Belirli Bir Pikselin Renk Kodunu Okuma

Fotoğraftaki y=200, x=300 koordinatındaki tek bir noktaya (piksele) mikroskopla bakalım. Hangi sayılardan oluşuyor?

```python
# Dikkat: Numpy matrisleri her zaman önce Y eksenini (Satır/Yükseklik), sonra X eksenini (Sütun/Genişlik) okur.
tek_piksel = duzgun_resim[200, 300] 

print("(Y:200, X:300) koordinatındaki pikselin değerleri:")
print(f"Kırmızı Yoğunluğu: {tek_piksel[0]} / 255")
print(f"Yeşil Yoğunluğu: {tek_piksel[1]} / 255")
print(f"Mavi Yoğunluğu: {tek_piksel[2]} / 255")
```

## Adım 10: Dene ve Keşfet - Görüntüyü Kesme (Cropping)

Bir matrisi dilimlemek (Slicing) aslında fotoğrafı kırpmak demektir. Sadece kedinin yüzünün olduğu alanı kırpalım.

```python
# [Baslangic_Y:Bitis_Y, Baslangic_X:Bitis_X] 
# Kedinin yüzü tahmini olarak 50 ile 400. satırlar ve 300 ile 600. sütunlar arasında
kedi_yuzu = duzgun_resim[50:400, 300:600]

plt.imshow(kedi_yuzu)
plt.axis("off")
plt.title("Kırpılmış Fotoğraf")
plt.show()
```
