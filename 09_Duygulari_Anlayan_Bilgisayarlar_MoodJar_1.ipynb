## Adım 2: Teknik Terimler ve Sözlük

- Sentiment Analysis (Duygu Analizi): Verilen metnin taşıdığı duygusal tonu (pozitif, negatif veya nötr) tespit etme algoritması.
- Hugging Face: Yapay zeka geliştiricilerinin (Google, Meta, Microsoft dahil) kendi devasa modellerini açık kaynak (ücretsiz) olarak paylaştığı, dünyanın en büyük yapay zeka model kütüphanesidir.
- Transformers: Hugging Face tarafından geliştirilen ve dil modellerini kullanmamızı sağlayan Python paketinin adıdır.
- Pre-trained Model (Önceden Eğitilmiş Model): Günlerce devasa sunucularda veri yığınlarıyla daha önceden eğitilmiş ve kural çıkarmış, kullanımı hazır modeldir.
- Pipeline (Veri Hattı): Bir metni tokenlarına (sayılara) ayırma, modele sokma ve sonucu bize anlama dönüştürme işlemlerinin tamamını arka planda otomatik yapan 'kestirme' araçtır.

## Adım 5: Önceden Eğitilmiş (Pre-trained) Modeli İndirme

Pipeline'a hangi görevi yapmak istediğimizi söylüyoruz. Biz "sentiment-analysis" (duygu analizi) yapmak istiyoruz. 
Not: Bu hücreyi ilk çalıştırdığınızda model dosyaları (yaklaşık 250 MB) internetten Colab sunucusuna indirilecektir, birkaç saniye sürebilir.

```python
# pipeline("görev_adi") şeklinde kullanılır.
# Varsayılan olarak İngilizce eğitilmiş bir model indirecektir.
duygu_modeli = pipeline("sentiment-analysis")

print("\nYapay Zeka modeli başarıyla indirildi ve kullanıma hazır.")
```

## Adım 6: İlk Duygu Analizi Tahmini

Modelimiz hafızasında. Şimdi ona bir cümle verelim ve sonucu görelim.

```python
# Test cümlesi (İngilizce)
ornek_cumle = "I am so happy to learn artificial intelligence today!"

# Modeli tıpkı bir fonksiyon çağırır gibi kullanıyoruz
sonuc = duygu_modeli(ornek_cumle)

print("Modelin Analiz Sonucu:")
print(sonuc)
```

## Adım 7: Çıktıdaki Liste ve Sözlük Yapısını Ayıklama

Yukarıdaki çıktıyı incelediğinizde köşeli parantez (Liste) içinde süslü parantez (Sözlük) olduğunu göreceksiniz. `[{'label': 'POSITIVE', 'score': 0.999}]`
- label: Bulunan duygu durumu
- score: Modelin bulduğu karardan ne kadar emin olduğu (0 ile 1 arası olasılık).

Uygulamamızın çalışması için bu karmaşık metinden sadece etiketi ve skoru ayıklamamız lazım.

```python
# Listenin ilk elemanını (0. indeks) alıyoruz
ilk_sonuc = sonuc[0]

# Sözlük içindeki 'label' ve 'score' anahtarlarını çekiyoruz
duygu_etiketi = ilk_sonuc['label']
eminlik_orani = ilk_sonuc['score']

print("Ayrıştırılmış Etiket (Duygu):", duygu_etiketi)
# Skoru okunaklı hale getirmek için round() fonksiyonuyla 3 basamağa yuvarlıyoruz.
print("Ayrıştırılmış Eminlik Skoru:", round(eminlik_orani, 3))
```

## Adım 8: Dinamik Bir His Kavanozu Fonksiyonu Yazma

Önceki derslerde öğrendiğimiz `def` (fonksiyon), `if-else` (karar mekanizması) ve `input()` bilgilerini birleştirerek MoodJar projesinin çekirdek motorunu yazalım.

```python
def kavanoza_at(kullanici_mesaji):
    # Mesajı modele gönder
    analiz_sonucu = duygu_modeli(kullanici_mesaji)
    etiket = analiz_sonucu[0]['label']
    
    print(f"\n--- [ {etiket} ] tespit edildi ---")
    
    # Uygulamanın zeka bölümü (Duyguya göre farklı tepki verme)
    if etiket == "NEGATIVE":
        print("Görünüşe göre biraz zor bir gün geçiriyorsun. ")
        print("Şunu unutma: Her fırtınanın ardından güneş doğar. Derin bir nefes al ve yola devam et.")
    else:
        print("Harika! İçindeki bu pozitif enerji çok değerli. ")
        print("Bu güzel hislerin gün boyunca seninle kalmasını dilerim!")
        
# Fonksiyonu farklı İngilizce metinlerle test ediyoruz.
kavanoza_at("I failed my math exam today, I feel terrible.")
kavanoza_at("The weather is amazing and I bought a new game!")
```

## Adım 9: Threshold (Eşik Değeri) Kontrolü

Yapay zeka her zaman haklı değildir. Bazen karmaşık (hem olumlu hem olumsuz) kelimelerin olduğu cümleleri anlamaz ve kararsız kalır. Kararsız kaldığını, `score` değerinin düşük olmasından (Örn: 0.55) anlarız. Buna müdahale etmeliyiz.

```python
karmasik_cumle = "The weather is bad but I am drinking a very delicious coffee in my warm house."

analiz = duygu_modeli(karmasik_cumle)
skor = analiz[0]['score']

print(f"Karmaşık cümle için modelin eminlik oranı: {skor:.3f}")

# Güvenlik duvarı (Eşik değeri): Eğer model %70'ten düşük bir eminliğe sahipse cevap vermesin.
if skor < 0.70:
    print("Sistem: Bu cümlede karışık duygular seziyorum. Sana net bir tavsiye veremem, biraz daha açık anlatır mısın?")
else:
    print("Sistem: Duygunu net olarak anladım.")
```

## Adım 10: Dene ve Keşfet - Çoklu Cümle Analizi ve Hız

Pipeline'lar, tek bir metin yerine koca bir listeyi de kabul edebilir. Aşağıdaki kod ile birden fazla hissi kavanoza aynı anda atıp modelin gücünü görün.

```python
hisler_listesi = [
    "I love this new laptop.",
    "This food tastes disgusting.",
    "I am so bored.",
    "Artificial intelligence is fascinating."
]

# Tek bir değişkene veriyoruz, liste olarak gönderiyoruz
toplu_sonuclar = duygu_modeli(hisler_listesi)

# For döngüsü ile sonuçları güzelce yazdırıyoruz
for metin, sonuc in zip(hisler_listesi, toplu_sonuclar):
    print(f"Metin: {metin} -> Duygu: {sonuc['label']}")
```

## Adım 11: Dene ve Keşfet - Dil Sınırlarını Zorlamak

Yüklediğimiz bu varsayılan model İngilizce metinlerle eğitilmiştir. Peki Türkçe cümle verirsek ne olur? Model Türkçe bilir mi?

```python
# Görev: Aşağıdaki Türkçe cümleyi modele verin ve sonucu görün.
turkce_cumle = "Bugün her şey çok ters gitti, çok üzgünüm."

turkce_sonuc = duygu_modeli(turkce_cumle)
print("Türkçe Deneme Sonucu:", turkce_sonuc)

# Not: Çıkan sonuç muhtemelen hatalıdır, çünkü model Türkçe 'üzgün' kelimesinin ne anlama geldiğini bilmiyor.
# Bir sonraki derste modelimizi tamamen Türkçe'ye uyarlayacağız!
```
