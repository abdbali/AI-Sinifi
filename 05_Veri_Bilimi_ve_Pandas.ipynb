## Adım 1: Teorik Çerçeve ve Günlük Hayat Analojisi

Yapay zeka (AI) ve makine öğrenmesi (ML) algoritmalarının yegane yakıtı 'veri'dir (data). İster bir araba galerisinin fiyat tahmin algoritmasını geliştirin, ister hastaların röntgenlerinden hastalık tespiti yapın, elinizde analiz edilecek düzenli veriler olmak zorundadır. Ancak Python'da devasa Excel dosyalarıyla (milyonlarca satır) saniyeler içinde başa çıkabilmek ve yapay zeka modellerine bu verileri aktarabilmek için özel bir araca ihtiyacımız vardır. Bu aracın adı 'Pandas' kütüphanesidir. Pandas, Python'un içindeki süper güçlendirilmiş, kodla yönetilen bir Excel programı gibi düşünülebilir.

## Adım 2: Teknik Terimler ve Sözlük

- Pandas: Python'da veri manipülasyonu ve analizi için en çok kullanılan açık kaynaklı kütüphane.
- DataFrame (Veri Çerçevesi): Pandas içindeki satır ve sütunlardan oluşan iki boyutlu tablo yapısıdır (Tıpkı bir Excel sayfası gibi).
- Series (Seri): DataFrame içerisindeki tek bir sütuna verilen isimdir.
- Dataset (Veri Seti): Analiz edilmek üzere toplanmış yapılandırılmış bilgi bütünü.
- Import (İçe Aktarma): Python'da dışarıdan bir kütüphaneyi kodumuzun içine dahil etme komutudur.
- Dictionary (Sözlük): Python'da verileri 'anahtar-değer' (key-value) eşleşmesiyle tutan özel bir veri yapısıdır.

## Adım 3: Sözlük (Dictionary) Veri Yapısı

Pandas veri çerçevelerini oluşturmadan önce, Python'daki 'sözlük' veri yapısını bilmeliyiz. Sözlükler süslü parantez '{}' ile oluşturulur ve İngilizce sözlüklerdeki gibi bir kelime (anahtar) ile karşılığındaki anlama (değer) sahiptir.

```python
# Ogrenci bilgilerini tutan bir sözlük (Dictionary)
ogrenci_bilgileri = {
    "İsim": "Ahmet",
    "Yaş": 16,
    "Not Ortalaması": 85.5
}

print("Öğrencinin İsmi:", ogrenci_bilgileri["İsim"])
print("Öğrencinin Yaşı:", ogrenci_bilgileri["Yaş"])

# Bir anahtara karşılık birden çok veri (Liste) de tutabiliriz
okul_verisi = {
    "İsimler": ["Ahmet", "Ayşe", "Veli"],
    "Yaşlar": [16, 17, 15]
}

print("\nTüm isimler:", okul_verisi["İsimler"])
```

## Adım 4: Pandas Kütüphanesini İçe Aktarma

Pandas, Python'un standart bir parçası değildir (Colab'de hazır yüklü gelir ancak kodumuza davet etmemiz gerekir). Bunun için `import` komutunu kullanırız.

```python
# Pandas kütüphanesini koda dahil ediyoruz ve her seferinde uzun uzun 'pandas' 
# yazmamak için ona 'pd' kısaltmasını (as pd) veriyoruz.
import pandas as pd

print("Pandas kütüphanesi başarıyla yüklendi.")
```

## Adım 5: Kendi Veri Çerçevemizi (DataFrame) Yaratmak

Şimdi öğrendiğimiz sözlük yapısını, Pandas DataFrame'ine çevirerek profesyonel bir tablo görünümü elde edeceğiz.

```python
veri_sozlugu = {
    "İsim": ["Ali", "Ayşe", "Mehmet", "Zeynep", "Can"],
    "Yaş": [25, 30, 22, 28, 35],
    "Meslek": ["Mühendis", "Doktor", "Öğrenci", "Avukat", "Mimar"],
    "Maaş": [45000, 60000, 0, 50000, 55000]
}

# pd.DataFrame() komutu, basit sözlüğümüzü gelişmiş bir analiz tablosuna çevirir.
df = pd.DataFrame(veri_sozlugu)

# Jupyter/Colab ortamında df yazıp bırakırsak güzel bir tablo olarak çizer.
df
```

## Adım 6: Veri Setine Genel Bakış (head ve info)

Bazen elimizde 100.000 satırlık veriler olur. Hepsini ekrana yazdırmak bilgisayarı dondurabilir. Sadece ilk birkaç satırı görmek veya tablonun yapısı hakkında bilgi almak için bazı metotlar kullanırız.

```python
# İlk 3 satırı gösterir (Varsayılanı 5'tir)
print("--- İlk 3 Satır ---")
print(df.head(3))

print("\n--- Tablo Bilgileri ---")
# Sütunların veri tiplerini ve boş (null) değer olup olmadığını gösterir
df.info()
```

## Adım 7: Belirli Bir Sütunu (Series) Seçmek

Tüm tablo yerine sadece ilgilendiğimiz bir sütunu (örneğin sadece Maaş verilerini) çekip alabiliriz.

```python
# Köşeli parantez ve sütun ismi ile seçim yapılır.
maaslar = df["Maaş"]

print("Sadece Maaş Sütunu:")
print(maaslar)

print("\nBu sütunun veri tipi:", type(maaslar))
```

## Adım 8: Koşullu Veri Filtreleme

Veri biliminin en büyük gücü filtrelemedir. Tıpkı `if` komutuna benzer bir mantıkla, sadece belirli bir şartı sağlayan satırları filtreleyebiliriz.

```python
# Sadece yaşı 25'ten büyük olanları filtreleyelim.
# Okunuşu: df'in içinden, df'in Yaş sütunu 25'ten büyük olan satırları getir.
buyuk_yaslilar = df[df["Yaş"] > 25]

print("Yaşı 25'ten Büyük Olan Kişiler:")
print(buyuk_yaslilar)

# İkinci bir örnek: Sadece Mühendis olanları bul
muhendisler = df[df["Meslek"] == "Mühendis"]
print("\nSadece Mühendisler:")
print(muhendisler)
```

## Adım 9: Yeni Bir Sütun Eklemek

Tabloya matematıksel hesaplamalarla kolayca yeni veri sütunları ekleyebiliriz. Örneğin maaşı yıllık formata çevirelim.

```python
# Yeni sütun ismi 'Yıllık Maaş' olsun.
# Değeri, mevcut 'Maaş' sütununun 12 ile çarpılmış hali olacaktır.
df["Yıllık Maaş"] = df["Maaş"] * 12

print("Yeni Sütun Eklendikten Sonra:")
print(df)
```

## Adım 10: Temel İstatistiksel Hesaplamalar

Pandas, For döngülerine ihtiyaç duymadan binlerce satırlık verinin istatistiklerini anında verebilen fonksiyonlara sahiptir.

```python
# mean() -> Ortalama hesaplar
ortalama_yas = df["Yaş"].mean()
print(f"Grubun yaş ortalaması: {ortalama_yas}")

# max() ve min() -> En büyük ve en küçük değeri bulur
en_yuksek_maas = df["Maaş"].max()
print(f"Gruptaki en yüksek maaş: {en_yuksek_maas}")

# sum() -> Toplamı verir
toplam_yillik_maas_yuku = df["Yıllık Maaş"].sum()
print(f"Şirketin toplam yıllık maaş yükü: {toplam_yillik_maas_yuku}")
```
